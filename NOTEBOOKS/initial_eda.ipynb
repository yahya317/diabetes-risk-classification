```python
import pandas as pd

df = pd.read_csv('../Data/diabetes_binary_health_indicators_BRFSS2015.csv')
```

```python
df.head()
```

```python
target = 'Diabetes_binary'

# proportion of each class in the target variable
proportions = df[target].value_counts(normalize=True).sort_index(ascending=True)

# map class labels to their corresponding names
class_names = {0: 'No Diabetes', 1: 'Prediabetes or Diabetes'}

for i, prop in proportions.items():
    print(f"{class_names[i]}: {prop:.2%}")
```

```python
df.columns
```

```python
# proportion of sex in the dataset
sex_proportions = df['Sex'].value_counts(normalize=True)
sex_proportions
```

```python
# check if people with no diabetes have a lower BMI than people with diabetes or prediabetes / map class labels to their corresponding names
bmi_by_diabetes = df.groupby('Diabetes_binary')['BMI'].mean().rename(index=class_names)

print("Average BMI by Diabetes Status:")
print(bmi_by_diabetes)
```

```python
# do more men or women have diabetes on average? / map class labels to their corresponding names
diabetes_by_sex = df.groupby('Sex')['Diabetes_binary'].mean().rename(index={0: 'Female', 1: 'Male'})

print("Average Diabetes Rate by Sex:")
print(diabetes_by_sex)

```
